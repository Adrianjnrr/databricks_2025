```python
import numpy as np

a= np.array([[1, 0], [0, 1]])
b= np.array([[1, 2], [3, 4]])
result = np.matmul(a,b)
print(result)
```

```python
def generate_array(n):
    return np.tile([1,2,3], n)
print(generate_array(3))    
```

```python
x = np.zeros((10, 10), dtype = int)
x[0, :] = 1
x[-1, :] = 1
x[:, 0] = 1
x[:, -1] = 1
print(x)
```

```python
x = np.random.rand(3,5)
sum_all = x.sum()
row_sum = x.sum(axis = 1)
col_sum = x.sum(axis = 0)
print (x)
print('Sum of all Array :\n', sum_all)
print('Row sum :', row_sum)
print('Column sum :', col_sum)
```

```python
x = np.array([1,1,0,0], dtype = bool)
y = np.array([1,0,1,0], dtype = bool)
print(np.bitwise_and(x,y))
print(np.bitwise_or(x,y))
```

```python
x = np.array([1,2,0,0,4,0])
indx = np.nonzero(x)[0]
print(indx)
```

```python
x = np.random.rand(8,8)
max_value = x.max()
min_value = x.min()
print('Maximum Value :', max_value)
print('Minimum Value :', min_value)
```

```python
def metrics(n):
    i = np.arange(n)
    return i[:, None] + i[None, :]   
print(metrics(4))     
```

```python
def create_matrix(n):
    rows = np.arange(1, n + 1)   
    cols = np.arange(n)
    return rows[:, None] * cols
print(create_matrix(4))    
```
